# 一、ndarray
## 1.ndarray的特征
### 多维性

```python
import numpy as np
a = np.array(5) # 创建0维ndarry
print(a)
print(f'a的维度是：{a.ndim}') #查看a的维度
b = np.array([1,2,3]) # 创建1维ndarry
print(b)
print(f'b的维度是：{b.ndim}')
c = np.array([[1,2,3],[4,5,6]])
print(c)
print(f'c的维度是：{c.ndim}')
```

### 同质性

```python
a = np.array([1,'String'])  #此时 如果存在不同类型的元素，ndarray会将所有元素转换为字符串类型 因为数组内要同类型的元素
print(a)
b = np.array([1,5.5,9])
print(b)
```

## 2.naarray的属性

```python
arr = np.array([[1,2,3], [4,5,6]])
print(arr)
print(f"arr的形状是{arr.shape}")
print(f"arr的维度是{arr.ndim}")
print(f"arr的元素个数是{arr.size}")
print(f"arr的数据类型是{arr.dtype}")
print(f"arr的转置是{arr.T}")
```

## 3.ndarray的创建

```python
# 基础创建方法 dtype：改变属性
arr = np.array([1,2,3], dtype = np.float64)
print(arr)
```

```python
# copy
arr1 = np.copy(arr) # 元素跟原始的数组相同，但是不是一个数组
print(arr1)
arr1[0] = 8
print(arr1)
print(arr)
```

```python
# 预定义形状
# 全0 全1 未初始化 固定值
# 全0
arr = np.zeros((2,3), dtype = np.int32)  #两行三列全0矩阵
# 如果是一维：np.zeros(5)
print(arr)
print(arr.dtype)
```

```python
# 全1
arr = np.ones((2,3), dtype = np.int32)  #两行三列全1矩阵
print(arr)
print(arr.dtype)
```

```python
arr = np.empty((2,3)) # 两行三列 但是是全空 没有写入数值
print(arr)
```

```python
arr = np.full((3,4), 7)  # 指定全为一个数值的矩阵 用full方法
print(arr)
```

```python
arr1 = np.zeros_like(arr)  # 创建一个全0矩阵，形状与arr相同
print(arr1)
arr2 = np.ones_like(arr)  # 创建一个全1矩阵，形状与arr相同
print(arr2)
```
