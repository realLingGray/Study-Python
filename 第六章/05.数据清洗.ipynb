```python
import pandas as pd

# 1、读取数据
df = pd.read_csv("resource/sales_with_errors.csv")

# 2、数据清洗
# 缺失值的处理

# 1、删除缺失值
#df.isnull()  #查看缺失值的 返回一个表格，每个格子是True or False
#df.dropna()  # 删除缺失值所在的那一行
df_del = df.dropna(axis = 1)  #如果将axis = 1，那么删除的就是缺失值所在的那一列
# 注：dropna删除后是将删除后的结果返回回来，而不是直接修改源文件
df_del

# 2、填充缺失值
df.fillna("---")
# 使用上一行的数据来填充
df.ffill()
#使用下一行的数据来填充
df.bfill()

```

```python
# 重复值的处理
# 查看重复值
df.duplicated() # 直接调用的话是查看所有的列是否都一致，如果一致则重复
# 如果检测key的话 可以传递参数进去
df.duplicated(subset="订单号")

# 删除重复值 - 根据subset找重复值然后删除 keep选择 - first保留第一个 last保留最后一个 False全部删除
df.drop_duplicates(subset = "订单号", keep = "first")
```

```python
# 异常值的处理
# 查看异常值
df[df["单价"] < 0]

# 删除异常值
df.drop(df[df["单价"] < 0].index)

# 修复异常值
df["单价"]= df["单价"].abs()
df
```

```python
# 数据格式的处理
# 转化为str 然后使用str字符串的方法去replace
df["订单日期"] = df["订单日期"].str.replace("/", "-")
df
```
