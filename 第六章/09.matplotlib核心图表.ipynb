```python
from matplotlib.axes import Axes
import matplotlib.pyplot as plt
from IPython.core.pylabtools import figsize
from matplotlib import axes
from matplotlib.patches import bbox_artist

# 展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']

# 创建子图  其中 nrows是行 ncols是列 返回值是tuple figure是画布
# axis是子图数组
figure, axis = plt.subplots(nrows = 1, ncols = 2, figsize = (20, 6), dpi = 100)
# 图1：柱状图（世界石油储备） --> bar

countries = ["美国", "中国", "俄罗斯", "沙特阿拉伯", "伊朗"]
values = [198, 85, 67, 60, 50]

# 取第一个图
axis1:Axes = axis[0]


# 画柱状图
axis1.bar(countries, values, width = 0.5, color = "g")
# 设置标题
axis1.set_title("世界石油储备", fontsize = 18)
# 设置X轴标签
axis1.set_xlabel("国家", fontsize = 14)
# 设置y轴标签
axis1.set_ylabel("石油储备(亿桶)", fontsize = 14)
# 设置网格线
axis1.grid(linestyle= "--", alpha = 0.3)



# 画饼状图
axis2:Axes = axis[1]
countries2 =['印度', '中国', '美国', '印尼', '巴基斯坦', '尼日利亚', '巴西', '俄罗斯', '其他']
values = [14.51, 14.09, 3.4, 2.83, 2.51, 2.33, 2.12, 1.44, 20]
# 设置饼状图 ---> pie
axis2.pie(values, labels = countries2, autopct = '%.1f%%')
# 设置标题title
axis2.set_title("世界人口比例", fontsize = 18)
# 设置图例
axis2.legend(loc = "lower center", ncol = 5, bbox_to_anchor = (0.5, -0.05)) # 显示图例，ncol：每行显示图例的个数，bbox_to_anchor：图例位置
# 第一个是左右相对位置 第二个是上下相对位置


# 保存图片
plt.savefig("./data/01.png")
# figure.savefig("world_oil.png")
```
