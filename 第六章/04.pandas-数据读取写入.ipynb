```python
import pandas as pd
from numpy.ma.extras import column_stack

#读取数据
df1 = pd.read_csv("resource/sales_200_rows.csv")

df2 = pd.read_csv("resource/sales_200_rows.csv", usecols=["订单编号","日期","城市","商品类别", "数量", "单价_人民币元"])

# 数据处理
df2["总价"] = df2["数量"] * df2["单价_人民币元"]

# 将数据保存：to_csv  df ---> csv
df2.to_csv("resource/output_csv1.csv", index = False)
```

```python
# 数据选择
df2


# 1、选择列 单列
df2["订单编号"]

# 2、选择列 多列
df2[["订单编号" , "日期"]]
df1[["订单编号" , "日期", "商品名称"]]

# 选择行 - iloc 和 loc
# iloc ----> 基于行号选择行：  语法：df.iloc[start:stop:step]
df1.iloc[0:5:1] #不包含结束位置 注意：基于行号索引 而不是index
# loc ----> 基于index选择行：  语法：df.loc[start:stop:step]
df2.loc[1:6:1]
```

```python
# 数据过滤 ---> df[filter]
# 1、获取 销售数量>= 10 的订单数据 直接在df[]，[]内部写出筛选条件
df1[df1["数量"] >= 10]

# 配置项
#pd.set_option("display.max_rows", 100)


# 2、获取产品城市为成都的产品
df1[df1["城市"].isin(["成都"])]

# 3、单价在100-200之间的 范围匹配 可以用到between函数 - 默认包含边界（inclusive = "left"）
df1[df1["单价_人民币元"].between(100,200)]

# 4、销售数量>= 8 ， 单价在100-200之间的订单数据 ------> 如果是多个条件
# （并且：&  或者：|）
df1[(df1["数量"] >= 8) & (df1["单价_人民币元"].between(100,200))]

# 5、获取产品城市为：深圳/北京 类别为：数码/家居
df1[(df1["城市"].isin(["北京", "深圳"])) & (df1["商品类别"].isin(["数码", "家具"]))]
```
