```python
import pandas as pd
df = pd.read_csv("resource/sales_200_rows.csv", nrows = 20)
df["销售金额"] = df["单价_人民币元"] * df["数量"]


# 3.分组
# 3.1 根据 城市 分组 统计各个类别的订单数量
df.groupby("城市")["订单编号"].count()

# 3.2 根据 产品分类 分组 统计各个类别的 销售数量 之和
df.groupby("商品类别")["数量"].sum()

# 3.3 根据 产品分类 分组 统计各个类别的 销售金额 之和
df.groupby("商品类别")["销售金额"].sum()

# 3.4 根据 产品分类 分组 统计各个类别的 最低商品单价
df.groupby("商品类别")["单价_人民币元"].min()

# 3.5 根据 产品分类 分组 统计各个类别的 最高商品单价
df.groupby("商品类别")["单价_人民币元"].max()

# 3.6 根据 产品分类 分组 统计各个类别的 平均商品单价
df.groupby("商品类别")["单价_人民币元"].mean()

# 3.7 根据 产品分类 分组 统计各个类别的 最低、最高、平均商品单价
# 此时使用 agg 函数 将多个统计函数组合在一起 如果多个需求 则传入list
df.groupby("商品类别")["单价_人民币元"].agg(["mean", "max", "min"])

# 3.8 根据 产品分类 分组 统计各个类别的 平均单价、销售数量之和、销售金额之和
# 此时 不要用[]内传数据了 此时直接调用agg函数 参数为字典
df.groupby("商品类别").agg({"单价_人民币元":"mean", "数量":"sum", "销售金额":"sum"})

# 转成字典
df.groupby("商品类别").agg({"单价_人民币元":"mean", "数量":"sum", "销售金额":"sum"}).to_dict()
```
