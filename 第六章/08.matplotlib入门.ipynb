```python
import random

import matplotlib.pyplot as plt

# 展示中文字体为黑体
plt.rcParams['font.sans-serif'] = ['SimHei']

x = []
y_mars = []
y_moon = []

# 目标：绘制折线图
for i in range(1,25):
    x.append(i)
    y_mars.append(random.randint(1,100))
    y_moon.append(random.randint(1,100))

# 设置画布的大小 figsize 设置宽和高 ——》单位英寸
plt.figure(figsize=(10,5))

# 绘制折线图 -> 如果没有画布 会自动创建画布
plt.plot(x, y_mars, label="火星")
plt.plot(x, y_moon, label="月球")


# 设置折线图的详细信息
plt.title("星球气温变化折线图", fontsize=20)

# 设置x轴标签
plt.xlabel("时间")

# 设置y轴标签
plt.ylabel("温度")

# 设置x轴刻度 参数：列表
plt.xticks(x[::1])

# 设置x轴刻度 参数：列表
yticks = [i*10 for i in range(0,11)]
plt.yticks(yticks)

# 绘制网格线
plt.grid(ls = '-.', alpha = 0.4)

# 设置图例
plt.legend(loc = "upper left")

# 显示图表
plt.show()
```
