```python
import pandas as pd

# 创建DataFrame - 方式一
df1 = pd.DataFrame([
    {"姓名":"张三", "年龄":18, "性别":"男"},
    {"姓名":"李四", "年龄":19, "性别":"女"}
]
)
# 方式二
df2 = pd.DataFrame({
    "姓名": ["张三", "李四"],
    "年龄": [18, 19],
    "性别": ["男", "女"]
}
)
# 方式三
df3 = pd.DataFrame([
    ["张三",18, "男"],
    ["李四",19, "女"]
], columns=["姓名","年龄","性别"])
# 方式四
df4 = pd.DataFrame([
    ["张三", 80, 98, 88],
    ["李四", 75, 85, 90],
    ["王五", 60, 70, 80]
], columns=["姓名", "数学", "英语", "语文"], index = ["a","b","c"]
)
# DataFrame属性：注，属性不用加括号
df4.index.tolist() # 获取索引
df4.columns #拿到每一列的列名
df4.values.tolist() #获取属性值
df4.size #获取总格数
df4.dtypes #数据类型
df4.shape #获取维度
```
